### Dataset Overview
- Rows: 9994 
- Columns: 19
- Numerical columns: 5 and Categorical columns: 14
- Order Date and Ship Date are stored as object type and require conversion to datetime before time-based analysis



### Key Columns
- **Sales**: Revenue generated from each product sale
- **Profit**: Profit earned on each transaction after subtracting business costs. Can be positive or negative
- **Discount**: Discount applied to the sale
- **Quantity**: Number of units sold

### Initial Business Observation
- The dataset contains transaction-level retail sales records.
- Each row represents a product purchased within an order.
- Order ID is not unique because one order can contain multiple products.


### Data Quality Assessment


```python
df.isna().sum()
```

```python
df.duplicated().sum()
```

**Observation**
- No missing values were found in any column.
- No duplicate rows were found in dataset.

```python
df['Order Date']=df['Order Date'].str.replace('-','/', regex=True)
```

```python
df['Order Date']=pd.to_datetime(df['Order Date'],format='%m/%d/%Y')
```

```python
df['Ship Date']=df['Ship Date'].str.replace('-','/',regex=True)
```

```python
df['Ship Date']=pd.to_datetime(df['Ship Date'],format='%m/%d/%Y')
```

**Observation**
- 'Order Date' and 'Ship Date' were stored as the 'object' data types.
- The date seperator was inconsistent.
- The seperators were standardized to 'datetime' datatype.
